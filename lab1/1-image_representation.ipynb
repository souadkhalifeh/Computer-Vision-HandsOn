# Split Channels

```python
rgb_image1= cv.imread("../data/colors.jpg")
rgb_image1 = cv.cvtColor(rgb_image1, cv.COLOR_BGR2RGB)
print(rgb_image1.shape)
```

```python
(R, G, B) = cv.split(rgb_image1)
```

# Show the image

Create a window for each channel

```python
cv.namedWindow("Blue", cv.WINDOW_NORMAL)
cv.namedWindow("Green", cv.WINDOW_NORMAL)
cv.namedWindow("Red", cv.WINDOW_NORMAL)
```

Show the image

```python
cv.imshow("Blue", B)
cv.imshow("Green", G)
cv.imshow("Red", R)
```

Save the images

```python
cv.imwrite("channel_red.jpg", R)
cv.imwrite("channel_green.jpg", G)
cv.imwrite("channel_blue.jpg", B)
```

# Drawing 

```python
import numpy as np
import cv2 as cv
```

```python
canvas= np.zeros((300, 300, 3), dtype="uint8")
```

```python
green = (0, 255, 0)
cv.line(canvas, (0, 0), (300, 300), green)
cv.imshow("Canvas", canvas)
```

```python
red = (0, 0, 255)
cv.line(canvas, (300, 0), (0, 300), red)
cv.imshow("Canvas", canvas)
```

```python
cv.rectangle(canvas, (50, 200), (200, 255), red , 5)
cv.imshow("Canvas", canvas)
cv.waitKey(0)
```

```python
blue = (255, 0, 0)
cv.rectangle(canvas, (50, 200), (200, 255), blue , -1)
cv.imshow("Canvas", canvas)
cv.waitKey(0)
```

### Draw a circle

```python
canvas2= np.zeros((300, 300, 3), dtype="uint8")
#the center of the canvas
(centerX, centerY) = (canvas2.shape[1] // 2, canvas2.shape[0] // 2)
white = (255, 255, 255)
for r in range(0, 150, 25):
    cv.circle(canvas2, (centerX, centerY), r, white, 2)
cv.imshow("Canvas2", canvas2)
cv.waitKey(0)
```

Random centers

```python
for i in range(0,25):
    radius = np.random.randint(5, high=200)
    color = np.random.randint(0, high=200, size=(3,)).tolist()
    pt = np.random.randint(0, high=300, size=(2,)).tolist()
    cv.circle(canvas2, tuple(pt), radius, color, -1)

cv.imshow("Canvas2", canvas2)
cv.waitKey(0)    
```
