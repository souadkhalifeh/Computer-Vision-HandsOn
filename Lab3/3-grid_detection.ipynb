```python
import cv2 as cv
import numpy as np 
import matplotlib.pyplot as plt
```

```python
flat_chess = cv.imread("../data/flat_chessboard.png")
plt.imshow(flat_chess)
```

```python
found, corners = cv.findChessboardCorners(flat_chess, (7, 7))
if found:
    print("Chessboard corners found")
else:
    print("Chessboard corners not found")    
```

```python
corners.shape
```

```python
flat_chess_copy = flat_chess.copy()
cv.drawChessboardCorners(flat_chess_copy, (7, 7), corners, found)
plt.imshow(flat_chess_copy)
```
