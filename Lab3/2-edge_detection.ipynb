```python
import cv2 as cv 
import numpy as np 
import matplotlib.pyplot as plt
```

```python
img = cv.imread("../data/sammy_face.jpg")
plt_img = cv.cvtColor(img, cv.COLOR_BGR2RGB)
plt.imshow(plt_img)
```

```python
edges = cv.Canny(image=img, threshold1=127, threshold2=127)
edges
```

```python
plt.imshow(edges, cmap="gray")
```

# Choosing Threshold

```python
med_value= np.median(img)
lower = int(max(0, 0.7*med_value))
upper = int(min(255, 1.3*med_value))
```

```python
edges = cv.Canny(image=img, threshold1=lower, threshold2=upper)
plt.imshow(edges, cmap="gray")
```

Blur to reduc e noise

```python
blurred_image = cv.blur(img, ksize=(5, 5))
edges = cv.Canny(image=blurred_image, threshold1=lower, threshold2=upper)
plt.imshow(edges)
```
